```python
import pandas as pd
fraud_test_df=pd.read_csv('fraudTest.csv')
fraud_test_df.head()
```

```python
data_for_cut = fraud_test_df['amt']
import pandas as pd
bins = [0, 50, 200, 1000, data_for_cut.max() + 1]
category = pd.cut(data_for_cut, bins)
print(category)
```

```python
print(pd.Series(category).value_counts())
```

```python
bin_names = ['Small Amount', 'Medium Amount', 'Large Amount', 'Very Large Amount']
print(pd.cut(data_for_cut, bins, labels=bin_names))
```

```python
import numpy as np

data_for_equal_bins = fraud_test_df['city_pop']
print(pd.cut(data_for_equal_bins, 5, precision=0))
```

```python
from sklearn.preprocessing import MinMaxScaler

numerical_cols = ['amt', 'lat', 'long', 'city_pop', 'unix_time', 'merch_lat', 'merch_long']
```

```python
scaler = MinMaxScaler()

# Create a copy to avoid modifying the original DataFrame directly
fraud_test_df_normalized = fraud_test_df.copy()

# Apply scaler to the numerical columns
fraud_test_df_normalized[numerical_cols] = scaler.fit_transform(fraud_test_df_normalized[numerical_cols])

print("Fraud Test DataFrame (Normalized - First 5 rows):")
display(fraud_test_df_normalized.head())
```

```python
scaler = MinMaxScaler()

# Create a copy to avoid modifying the original DataFrame directly
fraud_train_df_normalized = fraud_train_df.copy()

# Apply scaler to the numerical columns
fraud_train_df_normalized[numerical_cols] = scaler.fit_transform(fraud_train_df_normalized[numerical_cols])

print("Fraud Train DataFrame (Normalized - First 5 rows):")
display(fraud_train_df_normalized.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Define numerical columns to standardize (same as for min-max scaling)
numerical_cols = ['amt', 'lat', 'long', 'city_pop', 'unix_time', 'merch_lat', 'merch_long']
```

```python
scaler_standard = StandardScaler()

# Create a copy to avoid modifying the original DataFrame directly
fraud_test_df_standardized = fraud_test_df.copy()

# Apply scaler to the numerical columns
fraud_test_df_standardized[numerical_cols] = scaler_standard.fit_transform(fraud_test_df_standardized[numerical_cols])

print("Fraud Test DataFrame (Standardized - First 5 rows):")
display(fraud_test_df_standardized.head())
```

```python
scaler_standard = StandardScaler()

# Create a copy to avoid modifying the original DataFrame directly
fraud_train_df_standardized = fraud_train_df.copy()

# Apply scaler to the numerical columns
fraud_train_df_standardized[numerical_cols] = scaler_standard.fit_transform(fraud_train_df_standardized[numerical_cols])

print("Fraud Train DataFrame (Standardized - First 5 rows):")
display(fraud_train_df_standardized.head())
```

```python
# Count fraud transactions in fraud_test_df
fraud_test_count = fraud_test_df['is_fraud'].sum()
print(f"Number of fraud transactions in fraud_test_df: {fraud_test_count}")
```

```python
# Count fraud transactions in fraud_train_df
fraud_train_count = fraud_train_df['is_fraud'].sum()
print(f"Number of fraud transactions in fraud_train_df: {fraud_train_count}")
```

```python
# Calculate total fraud transactions
total_fraud_count = fraud_test_count + fraud_train_count
print(f"Total number of fraud transactions across both datasets: {total_fraud_count}")
```
